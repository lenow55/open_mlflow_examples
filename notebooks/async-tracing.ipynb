```python
import mlflow
from mlflow import MlflowClient
from openai import AsyncOpenAI
from mlflow.entities import SpanType
```

```python
client = MlflowClient(tracking_uri="http://127.0.0.1:5801")

experiment_tags = {
    "project_name": "async-openai",
    "mlflow.note.content": "Асинхронный openai",
}

experiment = client.get_experiment_by_name(name="AsyncOpenAI")
if not experiment:
    experiment = client.create_experiment(name="AsyncOpenAI", tags=experiment_tags)
```

```python
client = AsyncOpenAI(
    api_key="api_key",
    base_url="http://localhost:8953/v1"
)
```

```python
@mlflow.trace(span_type=SpanType.LLM)
async def async_func(message: str):
    return await client.chat.completions.create(
        model="Qwen/Qwen3-8B-FP8",
        messages=[{"role": "user", "content": message}],
    )


await async_func("What is MLflow Tracing?")
```
